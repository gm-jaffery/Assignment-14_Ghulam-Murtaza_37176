```python
import pandas as pd

stu = pd.read_csv(r"C:\Users\Ghulam-Murtaza\Downloads\student_lifestyle_dataset.csv")
stu.head()
```

```python
print("Answer:", round(stu["study_hours"].mean(), 2))
```

```python
sleep_avg = stu.groupby("gender")["sleep_hours"].mean()
print(sleep_avg.round(2))
print()
print("Answer:", sleep_avg.idxmax(), "sleep more")
```

```python
corr = stu["screen_time"].corr(stu["sleep_hours"])
print("Correlation:", round(corr, 3))
print("Answer: almost no relationship (value is close to 0)")
```

```python
stu.groupby("gender")["internet_usage"].mean().round(2)
```
